This notebook shows the mathematical foundation of the Eye Aspect Ratio (EAR) used in our driver monitoring system.

```python
def eye_aspect_ratio(eye):
    A = distance.euclidean(eye[1], eye[5])
    B = distance.euclidean(eye[2], eye[4])
    C = distance.euclidean(eye[0], eye[3])
    ear = (A + B) / (2.0 * C)
    return ear
```

### Geometric Formula: EAR
The core of this project is based on the geometric relationship between facial landmarks. To distinguish between an open and closed eye, calculate the **Eye Aspect Ratio (EAR)** using Euclidean distances between six specific coordinates:

The formula is defined as:
$$EAR = \frac{||p_2 - p_6|| + ||p_3 - p_5||}{2 ||p_1 - p_4||}$$

Where:
* $p_1, \dots, p_6$ are 2D facial landmark location.
* The numerator estimates the **vertical** distance i.e height of the eye.
* The denominator estimates the **horizontal** distance i.e width of the eye.
* When the eye closes, the vertical distance approaches zero, causing the EAR to drop significantly.

### Temporal Dynamics & False Positive Mitigation
A critical challenge in drowsiness detection is distinguishing between a natural blink (typically 100ms - 400ms) and a micro-sleep event (>1 second). 

By implementing a **Temporal Threshold** ($T_{frames} = 48$), ensure the alarm is only triggered when the EAR remains below the threshold for approximately 1.6 seconds, which drastically reduces false alerts caused by normal physiological blinking.
